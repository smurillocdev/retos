- Leer imagen a color con OpenCV

```python
import cv2
from google.colab.patches import cv2_imshow

imagen = cv2.imread('colores.jpg', cv2.IMREAD_COLOR)

cv2_imshow(imagen)
```

- Imprimir forma (shape)

```python
forma = imagen.shape

print(forma)
```

- Extraer 5 zonas de interés homogeneas

```python
roi1 = imagen[37:47, 982:992]
roi2 = imagen[1030:1040, 270:280]
roi3 = imagen[755:765, 326:336]
roi4 = imagen[385:395, 1167:1177]
roi5 = imagen[676:686, 617:627]
```

- Extraer 1 zona de interes no homogenea

```python
roi6 = imagen[400:500, 900:1000]
```

- Dibujr recuadro en las regiones en la imagen original

```python
img_copia = imagen.copy()

cv2.rectangle(img_copia, (982, 37), (992, 47), (0,0,255), 2)
cv2.putText(img_copia, 'roi1', (982, 32), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,0,255), 2)

cv2.rectangle(img_copia, (270, 1030), (280, 1040), (0,0,255), 2)
cv2.putText(img_copia, 'roi2', (270, 1025), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,0,255), 2)

cv2.rectangle(img_copia, (326, 755), (336, 765), (0,0,255), 2)
cv2.putText(img_copia, 'roi3', (326, 750), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,0,255), 2)

cv2.rectangle(img_copia, (1167, 385), (1177, 395), (0,0,255), 2)
cv2.putText(img_copia, 'roi4', (1167, 380), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,0,255), 2)

cv2.rectangle(img_copia, (617, 676), (627, 686), (0,0,255), 2)
cv2.putText(img_copia, 'roi5', (617, 671), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,0,255), 2)

cv2.rectangle(img_copia, (900, 400), (1000, 500), (0,0,255), 2)
cv2.putText(img_copia, 'roi6', (900, 395), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,0,255), 2)

cv2_imshow(img_copia)
```

- Obtener media y desviación estandar de cada uno de los canales en cada región

```python
rois = [roi1, roi2, roi3, roi4, roi5, roi6]

for i, roi in enumerate(rois, start=1):
  media, std = cv2.meanStdDev(roi)
  print(f"\nRoi {i}")
  print("Media", media.flatten())
  print("Desviación estándar", std.flatten())
```

### Conclusión de Medias y Desviación Estándar

**Observaciones sobre las Medias:**

- Las medias muestran una fuerte correlación con el color visual percibido en cada región. Por ejemplo, roi 1 tiene una tonalidad roja dominante. roi2 y roi4, con medias bajas reflejan sus colores grisáceos y oscuros. roi3 presenta una predominancia ligeramente mayor en el canal rojo. roi5 muestra valores elevados y más equilibrados entre verde y azul, por lo que hay más ruido o una variación de color más compleja.

**Observaciones sobre la Desviación Estándar:**

- Las ROI homogéneas exhiben desviaciones estándar muy bajas, casi cero en algunos casos. Confirmando la uniformidad de los colores. Al igual que en la media, se nota que roi5 tiene un mayor ruido visual que lo que se veía anteriormente.

- Por otro lado, la no homogénea, muestra las desviaciones estándar más altas. Por la presencia de una variedad de colores dentro de esta región, abarcando tonos de piel, grises y cafés, por lo que hay una mayor dispersión de los valores de píxeles alrededor de la media.
